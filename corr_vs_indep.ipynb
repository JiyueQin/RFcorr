```python
import numpy as np 
import gstools as gs
import pandas as pd
import matplotlib.pyplot as plt
import time
```

Consider the bivariate field $Z(s) = [X(s), Y(s)], s\in Z$. In general, $\rho=0$ does not mean that the field $X$ and $Y$ are independent. However, if $Z(s)\overset{iid}{\sim}Z(1)$ and $Z(1)$ is bivariate Gaussian, then  $\rho=0$ means that $X$ and $Y$ are independent. 

```python
Z = np.random.normal(size = 50)

X = Z[0:-1]
Y = Z[1:]

#print(X, Y)

np.corrcoef(X, Y)
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(X, label="X", lw=1.2)
ax.plot(Y, label="Y", lw=1.2)
ax.set_xlabel("index")
ax.set_ylabel("value")
ax.legend()
fig.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4))
ax.scatter(X, Y, s=10, alpha=0.5)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("Scatter plot of Y vs X")
fig.tight_layout()
plt.show()
```

```python
Z = np.random.normal(size=50)
X = Z
Y = Z**2 - 1   # E[XY] = E[Z^3] = 0  -> uncorrelated, but Y is a function of X
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(X, label="X", lw=1.2)
ax.plot(Y, label="Y", lw=1.2)
ax.set_xlabel("index")
ax.set_ylabel("value")
ax.legend()
fig.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4))
ax.scatter(X, Y, s=10, alpha=0.5)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("Scatter plot of Y vs X")
fig.tight_layout()
plt.show()
```
