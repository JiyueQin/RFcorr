### Effect size

```python
import numpy as np

np.random.seed(0)

control = np.random.normal(50, 10, 30)
treatment = np.random.normal(55, 10, 30)

# Cohen's d
def cohens_d(x, y):
    nx, ny = len(x), len(y)
    pooled_sd = np.sqrt(((nx - 1)*np.var(x, ddof=1) + (ny - 1)*np.var(y, ddof=1)) / (nx + ny - 2))
    return (np.mean(y) - np.mean(x)) / pooled_sd

d = cohens_d(control, treatment)
print("Cohen's d:", d)
```

###  difference in difference

```python
import pandas as pd
import numpy as np

# create synthetic DiD dataset
data = pd.DataFrame({
    "group": ["control"]*20 + ["treatment"]*20,
    "pre":   np.random.normal(50, 5, 40),
})

# treatment jumps +5 after treatment
data["post"] = data["pre"] + np.where(data["group"]=="treatment", 5, 0) + np.random.normal(0, 2, 40)

# DiD estimator
did = (data.loc[data.group=="treatment", "post"].mean() -
       data.loc[data.group=="treatment", "pre"].mean()) - \
      (data.loc[data.group=="control", "post"].mean() -
       data.loc[data.group=="control", "pre"].mean())

print("DiD estimate:", did)
```

### device ID
return the device ID with the largest value of usage_in_minutes


```python
def find_max_usage_device(data):
    max_usage = -1
    max_device = None
    
    for entry in data:
        device_id, usage_str = entry.split(",")
        usage = int(usage_str)
        
        if usage > max_usage:
            max_usage = usage
            max_device = device_id
    
    return max_device

data = [
    "iqttt,0077",
    "obvhd,0093",
    "flohd,0075"
]
find_max_usage_device(data)
```

### Problem 2
Given an array of distinct integers a and an integer k (where 1 <=k<=len(a)), your task is to return the array of all possible subarrays of a obtained by removing k contiguous elements in a.
That is, you wish to obtain a subarray of a by removing the first k elements in a, another subarray of a by removing the next k elements in a, and so on.
Note that since you are removing k elements from a, you always obtain a subarray of length (len(a)-k), and there are (len(a)-k+1) such subarrays.
For a = [2, 4, 6, 8, 10] and k = 3, the output should be solution(a, k) = [[8, 10], [2, 10], [2,4]]

```python
# a is a list
def solution(a, k):
    n = len(a)
    res = []
    for start in range(n - k + 1):
        # elements before the removed block + elements after it
        sub = a[:start] + a[start + k:]
        res.append(sub)
    return res
solution([2,4,6,8,10], k = 3)
```

```python
# a is a numpy array
def solution(a, k):
    n = len(a)
    results = []

    for start in range(n - k + 1):
        # remove a[start : start+k]
        sub = np.concatenate((a[:start], a[start + k:]))
        results.append(sub)

    return results

solution(np.array([2,4,6,8,10]), k = 3)
```

### Bayes rule
Factory F: A, B. product meet the goal: X = 1
given: P(X=1|A), P(X = 1|B), P(F = A)
output: P(F = A|X = 0)

```python
def bayes(p1A, p1B, pA):
    pB = 1-pA
    p1 = (1-p1A)*pA + (1-p1B)*pB
    return (1-p1A)*pA/p1

bayes(98/100, 1950/2000, 1/3)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Data
X = np.array([0., 2., 3.]).reshape(-1, 1)  # shape (3, 1)
Y = np.array([1., 2., 1.])                 # shape (3,)

squared_errors = []
n = len(X)

for i in range(n):
    # Leave one out: remove the i-th sample from training
    X_train = np.delete(X, i, axis=0)
    Y_train = np.delete(Y, i)
    
    # The left-out test point
    X_test = X[i].reshape(1, -1)
    Y_test = Y[i]
    
    # Fit linear regression model on the remaining n-1 points
    model = LinearRegression()
    model.fit(X_train, Y_train)
    
    # Predict the left-out point
    y_pred = model.predict(X_test)[0]
    
    # Store squared error
    squared_errors.append((Y_test - y_pred) ** 2)

# LOOCV Mean Squared Error
mse_loocv = np.mean(squared_errors)
print("LOOCV MSE:", mse_loocv)



import numpy as np
from sklearn.linear_model import LinearRegression
# shortcut for OLS
def loocv_mse_shortcut(X, y):
    model = LinearRegression().fit(X, y)
    y_hat = model.predict(X)
    residuals = y - y_hat

    # compute hat matrix diagonal h_ii
    X_ = np.column_stack([np.ones(len(X)), X])  # add intercept
    H = X_ @ np.linalg.inv(X_.T @ X_) @ X_.T
    h = np.diag(H)

    loocv_errors = residuals / (1 - h)
    return np.mean(loocv_errors ** 2)
print("LOOCV MSE:",  loocv_mse_shortcut(X, Y))


```
